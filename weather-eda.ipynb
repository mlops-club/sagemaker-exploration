```python
NC_FPATH = "./ncddsupp-202506-obcounts.nc"

```

```python
import xarray as xr

# Open the NetCDF file
ds = xr.open_dataset(NC_FPATH)

# See what variables and dimensions are available
print(ds)

# Convert to pandas DataFrame (optional: select variable and flatten dimensions)
df = ds.to_dataframe().reset_index()

print(df.head())
```

```python
# Configure pandas display options to avoid scientific notation
import pandas as pd
pd.set_option('display.float_format', '{:.2f}'.format)
pd.set_option('display.precision', 2)
pd.set_option('display.max_columns', None)  # Show all columns
pd.set_option('display.width', None)  # Auto-detect width
```

```python
df.describe()
```

```python
# Let's examine the dataset structure and variable names
print("Dataset variables and their attributes:")
for var_name in ds.data_vars:
    print(f"\n{var_name}:")
    if hasattr(ds[var_name], 'long_name'):
        print(f"  Long name: {ds[var_name].long_name}")
    if hasattr(ds[var_name], 'units'):
        print(f"  Units: {ds[var_name].units}")
    if hasattr(ds[var_name], 'description'):
        print(f"  Description: {ds[var_name].description}")
    print(f"  Shape: {ds[var_name].shape}")
    print(f"  Data type: {ds[var_name].dtype}")

print("\nDataFrame column names:")
print(df.columns.tolist())
```
